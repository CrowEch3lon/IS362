## Dataset 1: World Bank Infant Mortality Rate by Country (1960–2023)

This dataset from the World Bank's World Development Indicators contains the infant mortality rate 
(deaths per 1,000 live births) for countries and regional aggregates from 1960 to 2023.

The raw file is in **wide format**: each row is a country, and each year from 1960 onward is its own column. 
We need to skip 4 metadata rows at the top, drop redundant indicator columns, and melt the year 
columns into a tidy long format where each row is one country/year observation.

```python
df_imr = pd.read_csv('InfantMortalityRate.csv', skiprows=4)
df_imr.head()
```

## Inspection

Let's check the shape, column types, and whether any nulls exist

```python
df_imr.shape
```

```python
df_imr.dtypes
```

```python
df_imr.isnull().sum().head(10)
```

## Cleaning

The `Indicator Name` and `Indicator Code` columns are identical for every row (this dataset covers only one indicator), so we'll drop them. 

Then we'll melt all year columns into a single `Year` column to produce a tidy long format. 

Finally, we'll drop rows where the infant mortality 
value is missing, cast `Year` to integer, and filter out regional aggregates so we're working with individual countries only.

```python
df_imr = df_imr.drop(columns=['Indicator Name', 'Indicator Code'])

year_cols = [col for col in df_imr.columns if col.strip().isdigit()]

df_imr_long = df_imr.melt(
    id_vars=['Country Name', 'Country Code'],
    value_vars=year_cols,
    var_name='Year',
    value_name='Infant Mortality Rate (per 1,000)'
)

df_imr_long['Year'] = df_imr_long['Year'].astype(int)
df_imr_long = df_imr_long.dropna(subset=['Infant Mortality Rate (per 1,000)'])
```

```python
df_imr_long = df_imr_long[df_imr_long['Country Code'].str.len() == 3]

df_imr_long.head(10)
```

## Analysis

We'll look at the overall trend in life expectancy over time, then compare 
life expectancy between racial groups for both sexes combined.

```python
countries = ['United States', 'China', 'India', 'Brazil', 'Nigeria', 'Germany']
df_subset = df_imr_long[df_imr_long['Country Name'].isin(countries)]

plt.figure(figsize=(12, 5))
for country, group in df_subset.groupby('Country Name'):
    plt.plot(group['Year'], group['Infant Mortality Rate (per 1,000)'], label=country)

plt.title('Infant Mortality Rate Over Time — Selected Countries')
plt.xlabel('Year')
plt.ylabel('Deaths per 1,000 Live Births')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
latest_year = df_imr_long['Year'].max()
df_latest = df_imr_long[df_imr_long['Year'] == latest_year]

top10_worst = df_latest.nlargest(10, 'Infant Mortality Rate (per 1,000)')

plt.figure(figsize=(10, 5))
plt.barh(top10_worst['Country Name'], top10_worst['Infant Mortality Rate (per 1,000)'])
plt.xlabel('Deaths per 1,000 Live Births')
plt.title(f'Top 10 Countries by Infant Mortality Rate ({latest_year})')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

## Conclusions

Infant mortality has declined dramatically across all countries since 1960, reflecting global improvements in healthcare, sanitation, and nutrition.

Countries like China and Brazil show steep declines beginning in the 1980s–1990s, consistent with rapid economic development and healthcare investment during those periods.

India and Nigeria, while showing steady improvement, still have significantly higher rates than developed nations. The countries with the highest infant mortality in the most recent year are concentrated in Sub-Saharan Africa, highlighting persistent global health inequalities.

## Dataset 2: NYC Population by Borough (1950–2040)

This dataset tracks the total population of New York City's five boroughs across census years from 1950 to 2020, with projections through 2040.

The data is in "wide" format. Each decade is its own column. We need to melt it into a tidy "long" format where each row is one borough/year combination. 

We'll also clean up the borough name whitespace and drop the 
"borough share of total" columns to keep things simple.

```python
df_nyc = pd.read_csv('NYCPopByBorough1950-2040.csv')
df_nyc.head()
```

## Cleaning

First, we'll drop the "share of NYC total" columns since we only need raw population. 
Then we'll melt the year columns into a single `Year` column, and strip 
the leading whitespace from borough names.

### Keep only the plain population columns (drop the "Boro share" columns)

```python
pop_cols = ['Age Group', 'Borough'] + [str(y) for y in range(1950, 2050, 10)]
df_nyc = df_nyc[pop_cols]

df_nyc.head()
```

```python
df_nyc_long = df_nyc.melt(
    id_vars=['Age Group', 'Borough'],
    var_name='Year',
    value_name='Population'
)

df_nyc_long['Year'] = df_nyc_long['Year'].astype(int)
df_nyc_long['Borough'] = df_nyc_long['Borough'].str.strip()
df_nyc_long['Population'] = df_nyc_long['Population'].astype(int)

df_nyc_long.head(10)
```

## Analysis

Let's look at how each borough's population has changed over time, and compare their share of total NYC population.

```python
boroughs = df_nyc_long[df_nyc_long['Borough'] != 'NYC Total']

plt.figure(figsize=(12, 5))
for borough, group in boroughs.groupby('Borough'):
    plt.plot(group['Year'], group['Population'], marker='o', label=borough)

plt.title('NYC Borough Populations Over Time (1950–2040)')
plt.xlabel('Year')
plt.ylabel('Population')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
nyc_total = df_nyc_long[df_nyc_long['Borough'] == 'NYC Total'][['Year', 'Population']].rename(columns={'Population': 'NYC Total'})
borough_only = boroughs.merge(nyc_total, on='Year')
borough_only['Share (%)'] = (borough_only['Population'] / borough_only['NYC Total'] * 100).round(2)

plt.figure(figsize=(12, 5))
for borough, group in borough_only.groupby('Borough'):
    plt.plot(group['Year'], group['Share (%)'], marker='o', label=borough)

plt.title('Each Borough as a Share of NYC Total Population (1950–2040)')
plt.xlabel('Year')
plt.ylabel('Share of NYC Population (%)')
plt.legend()
plt.tight_layout()
plt.show()
```


## Conclusions

Brooklyn is consistently the most populous borough throughout the entire period. 

Manhattan's share of the city's total population has declined steadily since 1950, while Queens has grown. 

The city as a whole dipped in population during the 1970s before recovering and is projected to continue growing through 2040.


## Dataset 3: World Bank GDP Per Capita (1960–2024)

This dataset from the World Bank's World Development Indicators contains 
GDP per capita (in current US dollars) for countries and regional aggregates from 1960 to 2024.

The raw file has metadata rows at the top that need to be skipped, and the year columns need to be melted into a tidy long format. We'll also drop redundant columns and handle missing values.

```python
df_gdp = pd.read_csv('WorldBankGDPPerCapitaByCountry.csv', skiprows=4)
df_gdp.head()
```

## Cleaning

We'll drop the `Indicator Name` and `Indicator Code` columns since they are 
identical for every row, then melt the year columns into a single `Year` column. 
Finally, we'll drop rows with missing GDP values.

```python
df_gdp = df_gdp.drop(columns=['Indicator Name', 'Indicator Code'])
df_gdp.head()
```

```python
year_cols = [col for col in df_gdp.columns if col.strip().isdigit()]

df_gdp_long = df_gdp.melt(
    id_vars=['Country Name', 'Country Code'],
    value_vars=year_cols,
    var_name='Year',
    value_name='GDP Per Capita (USD)'
)

df_gdp_long['Year'] = df_gdp_long['Year'].astype(int)
df_gdp_long = df_gdp_long.dropna(subset=['GDP Per Capita (USD)'])

df_gdp_long.head(10)
```

## Analysis

Let's compare GDP per capita over time for a handful of major economies, 
then look at the top 10 countries by most recent GDP per capita.

```python
countries = ['United States', 'China', 'Germany', 'Japan', 'India', 'Brazil']
df_subset = df_gdp_long[df_gdp_long['Country Name'].isin(countries)]

plt.figure(figsize=(12, 5))
for country, group in df_subset.groupby('Country Name'):
    plt.plot(group['Year'], group['GDP Per Capita (USD)'], label=country)

plt.title('GDP Per Capita Over Time — Selected Countries')
plt.xlabel('Year')
plt.ylabel('GDP Per Capita (Current USD)')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
latest_year = df_gdp_long['Year'].max()
df_latest = df_gdp_long[df_gdp_long['Year'] == latest_year]

df_latest = df_latest[df_latest['Country Code'].str.len() == 3]

top10 = df_latest.nlargest(10, 'GDP Per Capita (USD)')

plt.figure(figsize=(10, 5))
plt.barh(top10['Country Name'], top10['GDP Per Capita (USD)'])
plt.xlabel('GDP Per Capita (Current USD)')
plt.title(f'Top 10 Countries by GDP Per Capita ({latest_year})')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

## Conclusions

The U.S., Germany, and Japan have maintained high GDP per capita throughout the period, while China and India show dramatic upward trajectories starting in the 1990s–2000s. 

The top 10 wealthiest countries by GDP per capita are dominated by small, highly developed nations. 

The wide variation in values and the large amount of missing data for early years reflects both the diversity of global economies and the limitations of historical economic data collection.
